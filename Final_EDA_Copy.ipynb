```python
import pandas as pd
import matplotlib.pyplot as plt

# Assuming you have already loaded your DataFrame 'df'

# Calculate the mean of 'total reviews' for each 'genre'
genre_reviews_mean = df.groupby(['genre'], as_index=False)['total reviews'].mean()

# Create a pie chart
plt.figure(figsize=(5, 5))   #size of chart
plt.pie(genre_reviews_mean['total reviews'], labels=genre_reviews_mean['genre'], autopct='%1.1f%%', startangle=140)
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.

plt.title('Mean Total Reviews by Genre')

plt.show()
```
