```python
import numpy as np

# Assuming you have a list of prices
prices = [10.99, 20.99, 30.99, 40.99, 50.99, 60.99, 70.99, 80.99, 90.99, 100.99]

# Assuming you have a list of books with repeated prices
books = prices * 20

# Splitting the prices into ranges
price_ranges = np.array_split(books, 5)

# Printing the count of books within each price range
for i, price_range in enumerate(price_ranges):
    count = len(price_range)
    print(f"Price Range {i + 1}: {count} books")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Assuming you have a list of prices and publishers
prices = [10.99, 20.99, 30.99, 40.99, 50.99, 60.99, 70.99, 80.99, 90.99, 100.99]
publishers = ['Publisher A', 'Publisher B', 'Publisher C', 'Publisher D', 'Publisher E',
              'Publisher F', 'Publisher G', 'Publisher H', 'Publisher I', 'Publisher J']

# Assuming you have a list of books sold by publishers
books_sold = [100, 200, 150, 300, 250, 180, 220, 120, 280, 350]

# Creating a bar chart
x = np.arange(len(prices))
width = 0.35

fig, ax = plt.subplots()
rects = ax.bar(x, books_sold, width)

# Adding labels and title
ax.set_xlabel('Prices')
ax.set_ylabel('Number of Books Sold')
ax.set_title('Prices vs. Number of Books Sold by Publishers')

# Adding x-axis labels
ax.set_xticks(x)
ax.set_xticklabels(prices)

# Rotating x-axis labels for better visibility
plt.xticks(rotation=45)

# Displaying the count of books on top of each bar
for rect in rects:
    height = rect.get_height()
    ax.annotate(f'{height}', xy=(rect.get_x() + rect.get_width() / 2, height),
                xytext=(0, 3), textcoords="offset points",
                ha='center', va='bottom')

plt.show()
```
