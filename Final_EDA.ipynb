```python
import pandas as pd
import matplotlib.pyplot as plt

# Read the data
df = pd.read_excel('/content/book_sales.xlsx')

# Create a bar chart
plt.figure(figsize=(12,6))  # Adjust the figure size if needed
plt.bar(df['publisher.name'], df['daily average.units sold'])

# Label the axes
plt.xlabel('Publisher')
plt.ylabel('daily average.units sold')

# Show the plot
#plt.xticks(rotation=0)  # Rotate x-axis labels for better readability
plt.show()

```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Assuming you have already loaded your DataFrame 'df'

# Calculate the mean of 'total reviews' for each 'genre'
genre_reviews_mean = df.groupby(['genre'], as_index=False)['total reviews'].mean()

# Create a pie chart
plt.figure(figsize=(5, 5))   #size of chart
plt.pie(genre_reviews_mean['total reviews'], labels=genre_reviews_mean['genre'], autopct='%1.1f%%', startangle=140)
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.

plt.title('Mean Total Reviews by Genre')

plt.show()
```
